```python
# ============================================
# OFFLINE In-Context Learning Prompt Evaluation
# NO API REQUIRED | SINGLE CELL
# ============================================

# Task
TASK = "Classify the sentiment of the sentence as Positive or Negative."

# Test data
TEST_SENTENCES = [
    ("I love this product", "Positive"),
    ("This is the worst experience ever", "Negative"),
    ("The service was okay", "Neutral")
]

# -------------------------------
# Prompting Strategies
# -------------------------------

def zero_shot_prompt(sentence):
    return f"{TASK}\nSentence: {sentence}"

def one_shot_prompt(sentence):
    return f"""
{TASK}
Sentence: I hate this phone
Sentiment: Negative

Sentence: {sentence}
"""

def few_shot_prompt(sentence):
    return f"""
{TASK}
Sentence: The movie was amazing
Sentiment: Positive

Sentence: The food was terrible
Sentiment: Negative

Sentence: {sentence}
"""

def chain_of_thought_prompt(sentence):
    return f"""
{TASK}
Think step by step.

Sentence: {sentence}
"""

# -------------------------------
# Mock LLM (Simulation)
# -------------------------------

def mock_llm(prompt):
    prompt = prompt.lower()
    if any(word in prompt for word in ["love", "amazing", "enjoy"]):
        return "Positive"
    elif any(word in prompt for word in ["worst", "hate", "terrible"]):
        return "Negative"
    else:
        return "Neutral"

# -------------------------------
# Evaluation
# -------------------------------

def evaluate(strategy_func, name):
    print(f"\n==== {name} ====")
    correct = 0

    for sentence, true_label in TEST_SENTENCES:
        prompt = strategy_func(sentence)
        prediction = mock_llm(prompt)

        if prediction == true_label:
            correct += 1

        print(f"Sentence   : {sentence}")
        print(f"Prediction : {prediction}")
        print(f"Actual     : {true_label}")
        print("-" * 40)

    accuracy = correct / len(TEST_SENTENCES)
    print(f"Accuracy: {accuracy:.2f}")

# -------------------------------
# Run All Strategies
# -------------------------------

evaluate(zero_shot_prompt, "Zero-Shot Prompting")
evaluate(one_shot_prompt, "One-Shot Prompting")
evaluate(few_shot_prompt, "Few-Shot Prompting")
evaluate(chain_of_thought_prompt, "Chain-of-Thought Prompting")
```
